ELBOW METHOD

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for i in range(1, 10):
    km = KMeans(n_clusters=i, random_state=5)
    km.fit(x)
    wcss.append(km.inertia_)

# plot OUTSIDE loop
plt.figure(figsize=(4,4))
plt.plot(range(1,10), wcss, color="red", marker="8")
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kv = KMeans(n_clusters=5, random_state=0)
y_means = kv.fit_predict(x)

score = silhouette_score(x[:2000], kv.labels_[:2000])

x_plt = x[:, [0, 2]]

plt.scatter(x_plt[y_means==0,0],x_plt[y_means==0,1],s=50,color='red',label='cluster 1')
plt.scatter(x_plt[y_means==1,0],x_plt[y_means==1,1],s=50,color='blue',label='cluster 2')
plt.scatter(x_plt[y_means==2,0],x_plt[y_means==2,1],s=50,color='green',label='cluster 3')
plt.scatter(x_plt[y_means==3,0],x_plt[y_means==3,1],s=50,color='orange',label='cluster 4')
plt.scatter(x_plt[y_means==4,0],x_plt[y_means==4,1],s=50,color='green',label='cluster 5')

plt.title('KMeans clustering')
plt.xlabel('Balance')
plt.ylabel('Purchases')
plt.show()
```

DBSCAN-Density based Spatial Clustering of Application with Noise Method

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
```

```python
k=5
neighbors=NearestNeighbors(n_neighbors=k)
z=neighbors.fit(x_sample)
distance,indices=z.kneighbors(x_sample)
distance=np.sort(distance[:,k-1])
# Plot graph
plt.plot(distance)
plt.title('K-Distance Graph')
plt.xlabel('Data Points')
plt.ylabel('Distance')
plt.show()
```

```python
eps_value=4.5 #taken from graph that becomes high after 400 to 500 especially 450 sudden increase
db=DBSCAN(eps=eps_value,min_samples=5)
y=db.fit_predict(x_sample)
print("Cluster labels:", set(y))

import numpy as np

unique, counts = np.unique(y, return_counts=True)
print(dict(zip(unique, counts)))
```

```python
plt.figure(figsize=(6,4))

for i in set(y):
    if i == -1:
        plt.scatter(x_plot[y==i, 0], x_plot[y==i, 1],
                    s=30, color='black', label='Noise')
    else:
        plt.scatter(x_plot[y==i, 0], x_plot[y==i, 1],
                    s=30, label=f'Cluster {i}')

plt.title('DBSCAN Clustering')
plt.xlabel('Balance')
plt.ylabel('Payments')
plt.legend()
plt.show()
```
