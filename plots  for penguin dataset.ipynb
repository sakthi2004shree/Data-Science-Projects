```python
s=s.rename(columns={'sex':'Gender'})
```

```python
s.head()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Gender',data=s,color='green')
plt.title('count of gender')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
count=s['species'].value_counts()
count
```

```python
count=s['island'].value_counts()
count
```

```python
plt.figure(figsize=(12,3))
sns.barplot(x='Gender',y='island',data=s,color='red')
plt.title('variation in species')
plt.xlabel('island')
plt.ylabel('species')
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.histplot(s['body_mass_g'],bins=15,kde=True,color='brown')
plt.title("Combination of His and Kde analysis")
plt.xlabel('body weight of penguin')
plt.ylabel('y-axis')
plt.show()
```

```python
surv_in_island=s['island'].value_counts()
surv_in_island
```

```python
survive_island=s['island'].value_counts()
plt.figure(figsize=(10,4))
plt.pie(survive_island, labels=['Biscoe','Dream','Torgersen'], colors=['Red','Orange','Green'],autopct='%1.1f%%')
plt.title('penguins living island')
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.boxplot(x='body_mass_g',data=s)
plt.title('Calculation of body weight')
plt.xlabel('Weight')
plt.ylabel('species')
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.scatterplot(x='body_mass_g',y='flipper_length_mm',hue='island',style='Gender',data=s)
plt.title('penguins beak vs living place')
plt.xlabel('body weight')
plt.ylabel('wings of penguin to swim')
plt.legend(title='survive')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.scatterplot(x='bill_depth_mm',y='bill_length_mm',data=s)
plt.title('penguins beak vs living place')
plt.xlabel('beak depth')
plt.ylabel('beak length')
plt.legend(title='survival')
plt.show()              
```

```python
numerical=s[['bill_length_mm','bill_depth_mm','flipper_length_mm','body_mass_g',]].corr()

plt.figure(figsize=(5, 4))
sns.heatmap(numerical,annot=True,cmap='rainbow')
plt.title('Correlation of heatmap for penguin dataset')
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.violinplot(x='island',y='body_mass_g',hue='Gender',split=True,data=s,palette='rainbow')
plt.title('Violinplot survival Representation')
plt.legend(title='Gender')
plt.xlabel('Island')
plt.ylabel('species')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.kdeplot(data=s['body_mass_g'])
plt.title('Body weight Calculation')
plt.xlabel('Body weight')
plt.ylabel('island')
plt.show()
```

```python
sns.pairplot(s)
```

```python
sns.pairplot(s[['bill_length_mm','species','bill_depth_mm']],hue='bill_length_mm',palette='husl',plot_kws={'alpha':0.7})
plt.suptitle('Pairplot for penguin dataset', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.title('swarmplot representation')
plt.xlabel('island')
plt.ylabel('body_mass_g')sns.swarmplot(x='species',y='body_mass_g',data=s,hue='island',palette='rainbow',dodge=True)
plt.legend(title='island list')
plt.show()
```
